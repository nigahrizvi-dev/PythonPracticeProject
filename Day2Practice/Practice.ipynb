## How many passangers have survived?

```python
print("Survival counts:")
 
display(df["Survived"].value_counts())
 
print("Survival percentages:")
 
display(
    df["Survived"].value_counts(normalize=True) * 100
)
```

## How many males and females were there in the titanic? Calculating there percantage.

```python
print("No. of Passengers:")
display(df["Sex"].value_counts())
 
print("Percentage of genders present:")
 
display(
    df["Sex"].value_counts(normalize=True) * 100
)
```

## How many passangers age was > than 50?

```python
count = (df["Age"] > 50).sum()

print(count)
```

## Name of passangers of age>50:

```python
print("Males older than 50:")
print(df[(df["Age"] > 50) & (df["Sex"] == "male")]["Name"])

print("\nFemales older than 50:")
print(df[(df["Age"] > 50) & (df["Sex"] == "female")]["Name"])
```
